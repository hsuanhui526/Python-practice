```python
users = {}  # 初始資料庫為空

while True:
    choice = input("\n請選擇操作選項 (a 註冊, b 登入, c 退出) ? ")

    # --- 註冊 (a) ---
    if choice == "a":
        user = input("請輸入帳號 : ")
        if user in users:
            print("帳號已存在，請重新輸入！")
        else:
            pw = input("請輸入密碼 : ")
            users[user] = pw
            print("註冊成功！")

    # --- 登入 (b) ---
    elif choice == "b":
        user = input("請輸入帳號 : ")

        # 關鍵：先檢查帳號，存在才問密碼
        if user not in users:
            print("帳號不存在，請重新輸入！")
        else:
            pw = input("請輸入密碼 : ")
            if users[user] == pw:
                print("登入成功！")
            else:
                print("密碼錯誤！")

    # --- 退出 (c) ---
    elif choice == "c":
        print("系統已退出")
        break

    else:
        print("無效選項，請輸入 a, b 或 c")
```
